# Distance Measures

Distance measures quantify how far apart two points (or vectors) are in a given space. They are fundamental in machine learning, information retrieval, clustering, and many other fields.

In this notebook we explore the most common distance measures, build them from scratch, and visualize how they behave.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, FancyArrowPatch
from matplotlib.colors import Normalize
from matplotlib import cm
```

## Setup: Two sample points

We'll use two 2D points throughout this notebook so we can visualize the distances.

```python
a = np.array([1.0, 2.0])
b = np.array([4.0, 6.0])

print(f"Point A: {a}")
print(f"Point B: {b}")
```

---
## 1. Euclidean Distance (L2)

The straight-line distance between two points. This is the most intuitive distance measure.

$$d_{\text{euclidean}}(\mathbf{a}, \mathbf{b}) = \sqrt{\sum_{i=1}^{n}(a_i - b_i)^2}$$

```python
def euclidean_distance(x, y):
    """Compute Euclidean (L2) distance between two vectors."""
    return np.sqrt(np.sum((x - y) ** 2))

d_euc = euclidean_distance(a, b)
print(f"Euclidean distance: {d_euc:.4f}")
print(f"Verify with np.linalg.norm: {np.linalg.norm(a - b):.4f}")
```

```python
fig, ax = plt.subplots(1, 1, figsize=(5, 5))
ax.set_xlim(-1, 8)
ax.set_ylim(-1, 8)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.set_title(f'Euclidean Distance = {d_euc:.2f}')

# Points
ax.plot(*a, 'o', ms=10, color='#2196F3', zorder=5)
ax.plot(*b, 'o', ms=10, color='#F44336', zorder=5)
ax.annotate('A', a, textcoords='offset points', xytext=(-15, 10), fontsize=12, fontweight='bold')
ax.annotate('B', b, textcoords='offset points', xytext=(-15, 10), fontsize=12, fontweight='bold')

# Straight line
ax.plot([a[0], b[0]], [a[1], b[1]], '-', color='#4CAF50', lw=2.5, label='Euclidean path')
ax.legend()
plt.tight_layout()
plt.show()
```

---
## 2. Manhattan Distance (L1)

Also called **taxicab** or **city-block** distance. It measures distance along axes at right angles, like navigating a grid of streets.

$$d_{\text{manhattan}}(\mathbf{a}, \mathbf{b}) = \sum_{i=1}^{n} |a_i - b_i|$$

```python
def manhattan_distance(x, y):
    """Compute Manhattan (L1) distance between two vectors."""
    return np.sum(np.abs(x - y))

d_man = manhattan_distance(a, b)
print(f"Manhattan distance: {d_man:.4f}")
```

```python
fig, ax = plt.subplots(1, 1, figsize=(5, 5))
ax.set_xlim(-1, 8)
ax.set_ylim(-1, 8)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.set_title(f'Manhattan Distance = {d_man:.2f}')

ax.plot(*a, 'o', ms=10, color='#2196F3', zorder=5)
ax.plot(*b, 'o', ms=10, color='#F44336', zorder=5)
ax.annotate('A', a, textcoords='offset points', xytext=(-15, 10), fontsize=12, fontweight='bold')
ax.annotate('B', b, textcoords='offset points', xytext=(-15, 10), fontsize=12, fontweight='bold')

# L-shaped path (horizontal then vertical)
ax.plot([a[0], b[0]], [a[1], a[1]], '--', color='#FF9800', lw=2.5)
ax.plot([b[0], b[0]], [a[1], b[1]], '--', color='#FF9800', lw=2.5, label='Manhattan path')

# Also show Euclidean for comparison
ax.plot([a[0], b[0]], [a[1], b[1]], '-', color='#4CAF50', lw=1.5, alpha=0.5, label='Euclidean path')
ax.legend()
plt.tight_layout()
plt.show()
```

---
## 3. Cosine Distance

Cosine **similarity** measures the angle between two vectors, ignoring their magnitude. Cosine **distance** = 1 - cosine similarity.

$$\text{sim}_{\cos}(\mathbf{a}, \mathbf{b}) = \frac{\mathbf{a} \cdot \mathbf{b}}{\|\mathbf{a}\| \, \|\mathbf{b}\|}$$

$$d_{\cos}(\mathbf{a}, \mathbf{b}) = 1 - \text{sim}_{\cos}(\mathbf{a}, \mathbf{b})$$

This is widely used in NLP and recommendation systems where the direction of a vector matters more than its length.

```python
def cosine_similarity(x, y):
    """Compute cosine similarity between two vectors."""
    return np.dot(x, y) / (np.linalg.norm(x) * np.linalg.norm(y))

def cosine_distance(x, y):
    """Compute cosine distance (1 - cosine similarity)."""
    return 1.0 - cosine_similarity(x, y)

sim = cosine_similarity(a, b)
d_cos = cosine_distance(a, b)
print(f"Cosine similarity: {sim:.4f}")
print(f"Cosine distance:   {d_cos:.4f}")
print(f"Angle between vectors: {np.degrees(np.arccos(np.clip(sim, -1, 1))):.2f} degrees")
```

```python
fig, ax = plt.subplots(1, 1, figsize=(5, 5))
ax.set_xlim(-0.5, 5)
ax.set_ylim(-0.5, 7)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.set_title(f'Cosine Similarity = {sim:.4f}')

# Draw vectors from origin
origin = np.array([0, 0])
ax.annotate('', xy=a, xytext=origin, arrowprops=dict(arrowstyle='->', color='#2196F3', lw=2.5))
ax.annotate('', xy=b, xytext=origin, arrowprops=dict(arrowstyle='->', color='#F44336', lw=2.5))
ax.annotate('A', a, textcoords='offset points', xytext=(8, 0), fontsize=12, fontweight='bold', color='#2196F3')
ax.annotate('B', b, textcoords='offset points', xytext=(8, 0), fontsize=12, fontweight='bold', color='#F44336')

# Draw angle arc
angle_a = np.degrees(np.arctan2(a[1], a[0]))
angle_b = np.degrees(np.arctan2(b[1], b[0]))
from matplotlib.patches import Arc
arc = Arc((0, 0), 1.5, 1.5, angle=0, theta1=min(angle_a, angle_b), theta2=max(angle_a, angle_b), color='#9C27B0', lw=2)
ax.add_patch(arc)
ax.annotate(f'{np.degrees(np.arccos(np.clip(sim, -1, 1))):.1f}°', xy=(0.9, 1.2), fontsize=10, color='#9C27B0')

plt.tight_layout()
plt.show()
```

## 4. Minkowski Distance (Lp)

A generalization that includes both Euclidean and Manhattan as special cases.

$$d_{\text{minkowski}}(\mathbf{a}, \mathbf{b}) = \left(\sum_{i=1}^{n} |a_i - b_i|^p \right)^{1/p}$$

- **p = 1** : Manhattan distance
- **p = 2** : Euclidean distance
- **p -> inf** : Chebyshev distance (max absolute difference)

```python
def minkowski_distance(x, y, p):
    """Compute Minkowski distance of order p between two vectors."""
    return np.sum(np.abs(x - y) ** p) ** (1.0 / p)

for p in [1, 2, 3, 5, 10, 50]:
    d = minkowski_distance(a, b, p)
    print(f"p = {p:>2d}  ->  distance = {d:.4f}")

# Chebyshev (p -> infinity)
d_cheb = np.max(np.abs(a - b))
print(f"p -> inf  ->  distance = {d_cheb:.4f}  (Chebyshev)")
```

### Unit circles under different p-norms

The shape of the "unit circle" (all points at distance 1 from the origin) changes with p, giving us intuition for how each norm measures distance.

```python
fig, axes = plt.subplots(1, 4, figsize=(14, 3.5))

theta = np.linspace(0, 2 * np.pi, 1000)
p_values = [1, 2, 3, 10]
colors = ['#FF9800', '#4CAF50', '#2196F3', '#9C27B0']

for ax, p, color in zip(axes, p_values, colors):
    # Points on the unit circle for Lp norm
    x = np.cos(theta)
    y = np.sin(theta)
    # Normalize each (x, y) so that Lp norm = 1
    norms = (np.abs(x)**p + np.abs(y)**p)**(1.0/p)
    x_unit = x / norms
    y_unit = y / norms

    ax.fill(x_unit, y_unit, alpha=0.15, color=color)
    ax.plot(x_unit, y_unit, color=color, lw=2)
    ax.set_xlim(-1.5, 1.5)
    ax.set_ylim(-1.5, 1.5)
    ax.set_aspect('equal')
    ax.grid(True, alpha=0.3)
    label = {1: 'Manhattan', 2: 'Euclidean', 3: 'L3', 10: 'L10 (near Chebyshev)'}
    ax.set_title(f'p = {p} ({label[p]})')

plt.suptitle('Unit Circles Under Different Lp Norms', fontsize=13, y=1.02)
plt.tight_layout()
plt.show()
```

## 5. Comparison: Distance Heatmaps

Let's visualize how each distance measure perceives the "landscape" of distances from a fixed reference point.

```python
ref = np.array([5.0, 5.0])
x_grid = np.linspace(0, 10, 200)
y_grid = np.linspace(0, 10, 200)
X, Y = np.meshgrid(x_grid, y_grid)

# Compute distances from each grid point to the reference
d_euclidean = np.sqrt((X - ref[0])**2 + (Y - ref[1])**2)
d_manhattan = np.abs(X - ref[0]) + np.abs(Y - ref[1])
d_chebyshev = np.maximum(np.abs(X - ref[0]), np.abs(Y - ref[1]))

fig, axes = plt.subplots(1, 3, figsize=(14, 4.5))
titles = ['Euclidean (L2)', 'Manhattan (L1)', 'Chebyshev (L-inf)']
data = [d_euclidean, d_manhattan, d_chebyshev]

for ax, d, title in zip(axes, data, titles):
    im = ax.contourf(X, Y, d, levels=20, cmap='viridis')
    ax.contour(X, Y, d, levels=8, colors='white', linewidths=0.5, alpha=0.5)
    ax.plot(*ref, 'r*', ms=12, zorder=5)
    ax.set_title(title)
    ax.set_aspect('equal')
    plt.colorbar(im, ax=ax, shrink=0.8)

plt.suptitle('Distance Heatmaps from Reference Point (5, 5)', fontsize=13, y=1.02)
plt.tight_layout()
plt.show()
```

## Summary

| Measure | Formula | When to use |
|---|---|---|
| **Euclidean (L2)** | $\sqrt{\sum(a_i - b_i)^2}$ | Default choice; meaningful in low dimensions |
| **Manhattan (L1)** | $\sum \lvert a_i - b_i \rvert$ | High-dimensional or sparse data; robust to outliers |
| **Cosine** | $1 - \frac{a \cdot b}{\|a\|\|b\|}$ | Text/embeddings where direction matters more than magnitude |
| **Minkowski (Lp)** | $(\sum \lvert a_i - b_i \rvert^p)^{1/p}$ | Tunable generalization; p is a hyperparameter |
| **Chebyshev (L-inf)** | $\max \lvert a_i - b_i \rvert$ | When the worst-case dimension difference matters |
